```python
import pandas as pd

df = pd.read_csv("Curriculum_Industry_Skill_Alignment (1).csv")

print(df.head())
print(df.shape)
```

```python
print(df.columns)
print(df.info())
print(df.isnull().sum())
```

```python
import pandas as pd


df["student_id"] = df["Student_ID"].str.replace("S", "").astype("int64")



df["CGPA"] = df["CGPA"].fillna(df["CGPA"].median())
df["Projects_Completed"] = df["Projects_Completed"].fillna(df["Projects_Completed"].median())



df["gender_encoded"] = df["Gender"].map({
    "Male": 0,
    "Female": 1
}).astype("int64")



df["skill_count"] = (
    (df["Programming"] > 5).astype(int) +
    (df["DSA"] > 5).astype(int) +
    (df["DBMS"] > 5).astype(int) +
    (df["OS"] > 5).astype(int) +
    (df["CN"] > 5).astype(int) +
    (df["OOP"] > 5).astype(int)
).astype("int64")



df["is_placed"] = (
    df["Placement_Status"] == "Placed"
).astype("int64")



df["semester"] = df["Semester"].astype("int64")
df["cgpa"] = df["CGPA"].astype("float32")
df["programming"] = df["Programming"].astype("float32")


df["industry_readiness"] = df["Industry_Readiness"].map({
    "Not Ready": 0,
    "Ready": 1
}).astype("int64")
```

```python
df[
    [
        "student_id",
        "semester",
        "gender_encoded",
        "cgpa",
        "programming",
        "skill_count",
        "is_placed",
        "industry_readiness"
    ]
].head()
```

```python
base_time = pd.Timestamp(
    "2025-01-01",
    tz="UTC"
)


df["event_timestamp"] = (
    base_time +
    pd.to_timedelta(
        df["student_id"],
        unit="s"
    )
)


df["created_timestamp"] = (
    df["event_timestamp"] +
    pd.Timedelta(seconds=1)
)
```

```python
feature_df = df[
    [
        "student_id",
        "event_timestamp",
        "created_timestamp",
        "semester",
        "gender_encoded",
        "cgpa",
        "programming",
        "skill_count",
        "is_placed"
    ]
].copy()
```

```python
feature_store_yaml = """
project: curriculum_industry_skill_alignment


registry: data/registry.db


provider: local


offline_store:
  type: file


online_store:
  type: sqlite
  path: data/online_store.db
"""


with open(
    f"{repo_path}/feature_store.yaml",
    "w"
) as f:
    f.write(feature_store_yaml)
```

```python
feature_definition = '''
from datetime import timedelta

from feast import (
    Entity,
    FeatureView,
    FeatureService,
    Field,
    FileSource
)

from feast.types import (
    Float32,
    Int64
)

# -----------------------------
# ENTITY
# -----------------------------

student = Entity(
    name="student",
    join_keys=["student_id"],
    description="CSE student"
)

# -----------------------------
# DATA SOURCE
# -----------------------------

curriculum_source = FileSource(
    name="curriculum_source",
    path="data/curriculum_features.parquet",
    timestamp_field="event_timestamp",
    created_timestamp_column="created_timestamp"
)

# -----------------------------
# FEATURE VIEW
# -----------------------------

curriculum_feature_view = FeatureView(
    name="curriculum_features",
    entities=[student],

    ttl=timedelta(days=50000),

    schema=[
        Field(name="semester", dtype=Int64),
        Field(name="gender_encoded", dtype=Int64),
        Field(name="cgpa", dtype=Float32),
        Field(name="programming", dtype=Float32),
        Field(name="skill_count", dtype=Int64),
        Field(name="is_placed", dtype=Int64),
    ],

    source=curriculum_source,

    online=True
)

# -----------------------------
# FEATURE SERVICE
# -----------------------------

curriculum_feature_service = FeatureService(
    name="industry_readiness_service",
    features=[
        curriculum_feature_view
    ]
)
'''

with open(
    f"{repo_path}/features.py",
    "w"
) as f:
    f.write(feature_definition)
```
