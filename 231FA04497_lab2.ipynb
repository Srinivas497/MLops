```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.read_csv("/content/Curriculum_Industry_Skill_Alignment (1).csv")

print("Dataset shape:", data.shape)
data.head()
```

```python
data.info()
```

```python
print(data.isnull().sum())
```

```python
features = [
    "Gender",
    "Semester",
    "CGPA",
    "Programming",
    "DSA",
    "DBMS",
    "OS",
    "CN",
    "OOP",
    "Python",
    "Java",
    "SQL",
    "Web_Development",
    "Cloud",
    "AI_ML",
    "Communication",
    "Aptitude",
    "Teamwork",
    "Problem_Solving",
    "Internship",
    "Certifications",
    "Projects_Completed",
    "Placement_Status"
]

target = "Industry_Readiness"

X = data[features]
y = data[target]

print(X.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
numerical_only = [
    "Semester",
    "CGPA",
    "Programming",
    "DSA",
    "DBMS",
    "OS",
    "CN",
    "OOP",
    "Python",
    "Java",
    "SQL",
    "Web_Development",
    "Cloud",
    "AI_ML",
    "Communication",
    "Aptitude",
    "Teamwork",
    "Problem_Solving",
    "Certifications",
    "Projects_Completed"
]

baseline_train = X_train[numerical_only].copy()
baseline_test = X_test[numerical_only].copy()

train_valid_rows = baseline_train.dropna().index
test_valid_rows = baseline_test.dropna().index

baseline_train = baseline_train.loc[train_valid_rows]
baseline_test = baseline_test.loc[test_valid_rows]

y_train_baseline = y_train.loc[train_valid_rows]
y_test_baseline = y_test.loc[test_valid_rows]

baseline_model = LogisticRegression(max_iter=1000)

baseline_model.fit(baseline_train, y_train_baseline)

baseline_predictions = baseline_model.predict(baseline_test)

baseline_accuracy = accuracy_score(
    y_test_baseline,
    baseline_predictions
)

print("Baseline accuracy:", baseline_accuracy)
print("Training records used:", len(baseline_train))
print("Testing records used:", len(baseline_test))
```

```python
numerical_features = [
    "Semester",
    "CGPA",
    "Programming",
    "DSA",
    "DBMS",
    "OS",
    "CN",
    "OOP",
    "Python",
    "Java",
    "SQL",
    "Web_Development",
    "Cloud",
    "AI_ML",
    "Communication",
    "Aptitude",
    "Teamwork",
    "Problem_Solving",
    "Certifications",
    "Projects_Completed"
]

categorical_features = [
    "Gender",
    "Internship",
    "Placement_Status"
]
```

```python
numerical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)
```

```python
categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_pipeline,
            numerical_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)
```

```python
complete_pipeline = Pipeline(
    steps=[
        (
            "preprocessing",
            preprocessor
        ),
        (
            "model",
            LogisticRegression(max_iter=1000)
        )
    ]
)
```

```python
complete_pipeline.fit(X_train, y_train)
```

```python
processed_predictions = complete_pipeline.predict(X_test)

processed_accuracy = accuracy_score(
    y_test,
    processed_predictions
)

print("Accuracy after preprocessing:", processed_accuracy)
```

```python
print(
    classification_report(
        y_test,
        processed_predictions
    )
)
```

```python
comparison = pd.DataFrame(
    {
        "Approach": [
            "Before proper preprocessing",
            "After preprocessing pipeline"
        ],
        "Accuracy": [
            baseline_accuracy,
            processed_accuracy
        ],
        "Training records used": [
            len(baseline_train),
            len(X_train)
        ],
        "Testing records used": [
            len(baseline_test),
            len(X_test)
        ]
    }
)

print(comparison)
```

```python
X_train_processed = complete_pipeline.named_steps[
    "preprocessing"
].transform(X_train)

X_test_processed = complete_pipeline.named_steps[
    "preprocessing"
].transform(X_test)
```

```python
feature_names = complete_pipeline.named_steps[
    "preprocessing"
].get_feature_names_out()

print(feature_names)
```

```python
processed_train_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

processed_test_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)
```

```python
processed_train_df["Survived"] = y_train.values
processed_test_df["Survived"] = y_test.values
```

```python
processed_train_df.to_csv(
    "curriculum_processed_train.csv",
    index=False
)

processed_test_df.to_csv(
    "curriculum_processed_test.csv",
    index=False
)
```
